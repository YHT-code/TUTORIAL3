```python
import numpy as np
```

```python
array_2x2 = np.array([[1, 2], [3, 4]])
print(array_2x2)
```

```python
random_3x4 = np.random.rand(3, 4)
print(random_3x4)
```

```python
ones_array = np.ones((2, 3))
print("Array of ones:\n", ones_array)

# Zeroes array
zeros_array = np.zeros((2, 3))
print("Array of zeroes:\n", zeros_array)
```

```python
rand_6x3 = np.random.rand(6, 3)
print("Original 6x3 array:\n", rand_6x3)

transposed = rand_6x3.T
print("Transposed (3x6) array:\n", transposed)
```

```python
n, p, m = 3, 2, 4

# Random arrays
A = np.random.rand(n, p)
B = np.random.rand(p, m)

print("Matrix A:\n", A)
print("Matrix B:\n", B)

# Dot product
dot_product = np.dot(A, B)
print("Dot product (A·B):\n", dot_product)
```
